## 1. What is a String?

A **string** is a sequence of characters enclosed in quotes.

Python does not have a separate `char` data type. A single character is also a string of length 1.

```text
"RCB"
 ↑
string containing 3 characters
```


````python
team = "Royal Challengers Bengaluru"
player = 'Virat Kohli'
city = "Bengaluru"

print(team)
print(player)
print(city)

print(type(team))
print(len(team))
````

## 2. Creating Strings

Python supports single quotes, double quotes, and triple quotes.

Triple quotes are useful for multi-line text.


````python
team1 = 'RCB'
team2 = "CSK"

match_report = """RCB vs CSK
Venue: Bengaluru
Result: RCB won
"""

print(team1)
print(team2)
print(match_report)
````

## 3. String Indexing

Strings are indexed from **0**.

For `RCB`:

```text
R   C   B
0   1   2
```

Negative indexing starts from the right:

```text
R    C    B
-3  -2   -1
```


````python
team = "RCB"

print(team[0])
print(team[1])
print(team[-1])
print(team[-2])
````

## 4. String Slicing

Syntax:

```python
string[start:stop:step]
```

The `stop` index is **not included**.


````python
team = "Royal Challengers Bengaluru"

print(team[0:5])
print(team[:5])
print(team[6:])
print(team[::2])
print(team[::-1])
````

## 5. Strings are Immutable

You cannot directly change one character inside a string.

This will produce an error:

```python
team[0] = "X"
```

Instead, create a new string.


````python
team = "RCB"

# team[0] = "X"   # Uncomment to see TypeError

new_team = "X" + team[1:]
print(new_team)
````

## 6. Concatenation and Repetition

`+` joins strings.

`*` repeats a string.


````python
team = "RCB"
opponent = "CSK"

match = team + " vs " + opponent
print(match)

print("🏏 " * 3)
````

# String Searching and Checking

These methods are particularly useful when processing match descriptions, player records, comments, or user input.


## 7. `find()` and `rfind()`

- `find()` returns the first position.
- `rfind()` returns the last position.
- If the substring is absent, they return `-1`.


````python
match = "RCB defeated CSK and RCB qualified"

print(match.find("RCB"))
print(match.rfind("RCB"))
print(match.find("MI"))
````

## 8. `index()` and `rindex()`

They work similarly to `find()` and `rfind()`, but raise `ValueError` when the text is not found.


````python
match = "RCB defeated CSK"

print(match.index("CSK"))
print(match.rindex("RCB"))

# print(match.index("MI"))  # ValueError
````

## 9. `count()`

Counts how many times a substring occurs.


````python
commentary = "RCB RCB RCB! What a match."

print(commentary.count("RCB"))
print(commentary.count(" "))
````

## 10. `startswith()` and `endswith()`

Useful for checking prefixes and suffixes.


````python
team = "Royal Challengers Bengaluru"

print(team.startswith("Royal"))
print(team.startswith("RCB"))
print(team.endswith("Bengaluru"))
print(team.endswith("Mumbai"))
````

# Changing Case

Common methods:
- `upper()`
- `lower()`
- `capitalize()`
- `title()`
- `swapcase()`
- `casefold()`


````python
text = "royal challengers bengaluru"

print(text.upper())
print(text.lower())
print(text.capitalize())
print(text.title())

mixed = "RcB vS CsK"
print(mixed.swapcase())
````

## 11. Removing Whitespace

Methods:
- `strip()` — both sides
- `lstrip()` — left side
- `rstrip()` — right side


````python
team = "   RCB   "

print(repr(team))
print(repr(team.strip()))
print(repr(team.lstrip()))
print(repr(team.rstrip()))
````

## 12. `replace()`

Replace one piece of text with another.

Syntax:

```python
text.replace(old, new)
```

You can optionally specify how many replacements to perform.


````python
match = "RCB vs CSK"

print(match.replace("CSK", "MI"))

text = "RCB RCB RCB"
print(text.replace("RCB", "IPL", 1))
````

# Splitting and Joining

These are among the most important methods when processing real-world text.


## 13. `split()`

Converts a string into a list.

This is extremely useful when reading space-separated input.


````python
teams = "RCB CSK MI KKR"

team_list = teams.split()

print(team_list)
print(type(team_list))

match = "RCB,CSK,MI,KKR"
print(match.split(","))
````

## 14. `rsplit()`

Like `split()`, but starts splitting from the right.

The `maxsplit` argument controls the number of splits.


````python
text = "IPL 2026 RCB Bengaluru Final"

print(text.split())
print(text.rsplit(" ", 1))
````

## 15. `splitlines()`

Splits multi-line text into a list of lines.


````python
report = """RCB vs CSK
Venue: Bengaluru
Result: RCB won
Score: 185/4"""

print(report.splitlines())
````

## 16. `join()`

`join()` combines an iterable of strings.

Think of it as:

```text
list of strings → one string
```


````python
teams = ["RCB", "CSK", "MI", "KKR"]

print(", ".join(teams))
print(" vs ".join(["RCB", "CSK"]))
print("
".join(teams))
````

# Character Validation Methods

These methods return `True` or `False`.

Important methods:
- `isalnum()`
- `isalpha()`
- `isascii()`
- `isdecimal()`
- `isdigit()`
- `isnumeric()`
- `isidentifier()`
- `islower()`
- `isupper()`
- `istitle()`
- `isspace()`
- `isprintable()`


````python
samples = ["RCB2026", "RCB", "2026", "   ", "Royal Challengers", "RCB_2026"]

for value in samples:
    print("Value:", repr(value))
    print("isalnum :", value.isalnum())
    print("isalpha :", value.isalpha())
    print("isdigit :", value.isdigit())
    print("islower :", value.islower())
    print("isupper :", value.isupper())
    print("isspace :", value.isspace())
    print("isidentifier:", value.isidentifier())
    print()
````

````python
names = ["runs", "player_name", "player-name", "2026score", "score2026"]

for name in names:
    print(name, "->", name.isidentifier())
````

````python
print("Royal Challengers Bengaluru".istitle())
print("RCB".istitle())

print("RCB2026".isascii())
print("RCB\nCSK".isprintable())
print("RCB2026".isprintable())
````

# Formatting Strings

Three useful approaches:

1. f-strings — modern and usually preferred
2. `format()`
3. `%` formatting — older style, useful to recognize


````python
team = "RCB"
player = "Virat Kohli"
runs = 741

print(f"{player} scored {runs} runs for {team}.")
print(f"Team: {team}, Runs: {runs}, Average: {runs / 14:.2f}")
print("\thello")
print("hi\nolla")
````
