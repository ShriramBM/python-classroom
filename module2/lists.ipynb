## List
-  a list is a built-in data structure used to store an ordered, changeable (mutable) collection of items in a single variable
-  Lists are created using square brackets [] with items separated by commas
```python
    # syntax 
    variable_name = [] # empty list
    variable_name = [val1, val2, val3, .......] 
```

````python
# Nested List

matrix = [[1,2,3],[4,5,6],[7,8,9]]
print(matrix)

# if you want to write visually good 
matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

#how to access a element
print(matrix[1][1])
````

````python
# program to write matrix Element-Wise Addition usefull in Datascience
matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
e = 10
# Hint use loops
for i in range(0,len(matrix)):
    for j in range(0, len(matrix)):
        matrix[i][j] = matrix[i][j]+ e

print(matrix)
````
