## List
-  a list is a built-in data structure used to store an ordered, changeable (mutable) collection of items in a single variable
-  Lists are created using square brackets [] with items separated by commas
```python
    # syntax 
    variable_name = [] # empty list
    variable_name = [val1, val2, val3, .......] 
```

````python
# you can use in for loop to select each element in the list

ipl_teams = ["RCB", "MI", "CSK", "KKR", "SRH", "DC", "PBKS", "RR", "GT", "LSG"]
for team in ipl_teams:
    print(f"Welcome {team} for 2026 IPL season")
````
