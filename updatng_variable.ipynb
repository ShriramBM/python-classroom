## Assignment and Updating variable

```python
a = 10
b = 20
a = a + b
print(a)
```

```python
i,j = 100, 200
print(i)
print(j)
```

```python
x = 100
x = x + 1
print(x)
```

```python
y = 200
y += 1
print(y)
```

## Passing the value
- Pass by value
- Pass by Reference
- python's basic type like int, float , bool, string are pass my value
- python's objects are pass by reference like list, object, class 

```python
a = 10
b = a
print(b)
```

## Swap operations

```python
x = 100
y = 200
print("x value:",x)
print("y value:",y)
# do some magic and get x as 200 and y as 100
temp = x
x = y
y = temp

print("After swap x value is:",x)
print("After swap y value is:",y)
```

```python
x = 10
y = 20
x,y = y,x
print(x)
print(y)
```
