## Some conditional Operators 
- `>` ,`<`, `>=`, `<=`, `==` comparison operators
- `and` , `or` logical operator

```python
print(10>20) 
print(True or False)
print(True and False)

# program to check you are teenager or not
age = 18
if age < 20 and age > 12:
    print("You are a teenager")
```

## 1. Simple if statement 

```python
age = 20
if age >= 18:
    print("You are allowed for driver's licence")
```

## 2. if else statments

```python
age = 12
if age >= 18:
    print("You are allowed for driver's licence")
else:
    print("You are not allowed for driver's licence")
```

## 3. if elif else ladder

```python
marks = 67
if marks >=80 and marks <=100:
    print("A+")
elif marks >= 75 and marks <= 79:
    print("A")
elif marks >= 65 and marks <= 69:
    print("A-")
elif marks >= 60 and marks <= 64:
    print("B+")
elif marks >= 45 and marks <= 59:
    print("C+")
else:
    print("Failed")
```

## 4. match statments

```python
color = "red"
match color:
    case "blue":
        print("You have a bright future")
    case "red":
        print("You are sharp")
    case "yellow":
        print("you are great")
    case "pink":
        print("You are beautiful")
    case _:
        print("all are nice")
```

## 5. Nested if statments

```python
pin_correct = True
account_balance = 5000
withdrawal_amount = 6000

# Outer Condition: Security Check
if pin_correct:
    print("PIN verified successfully.")
    
    # Inner Condition: Financial Check
    if withdrawal_amount <= account_balance:
        print("Dispensing cash. Please take your money.")
    else:
        print("Transaction declined: Insufficient funds.") # This runs
        
else:
    print("Transaction declined: Invalid PIN entered.")
```

## 6. One line if statment

```python
age = 67
if age >= 60: print("You are allowed to Sit")
```

## 7. Ternary operation

```python
# without ternary operation
age = 34
status = ""
if age >= 18 :
    status = "adult"
else:
    status = "child"
print(status)
```

```python
age = 34
status = "adult" if age>=18 else "child"
print(status)
```
