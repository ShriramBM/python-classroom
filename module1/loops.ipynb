## For loop
- `for` statement iterates over the items of any sequence
```python
    for element in generator():
        # statments
```

````python
# example 1 with ranges function
for i in range(1, 11):
    print("hi ",i)
````

### Range Function
- range(from, to) `from <= i < to`
- range(to) it start `0 <= i < to`
- range(10, 0, -1) `10 >= i > 0`
- range(0,10,2) `0 <= i < 10` but incremented by 2 like i=i+2

````python
# multiplication table
for i in range(1, 11):
    print(f"2 * {i} = {2*i}") 
````

````python
# Break statments
for n in range(20):
    if n == 12:
        break
    print(n)
````

````python
# continue statments
for x in range(20):
    if(x%2 == 0):
        continue
    print(x)
````

````python
# pass statment
for x in range(100):
    pass
````

````python
# Nested Loop 
for i in range(1, 11):
    print(f"{i} tables")
    for j in range(1,11):
        print(f"{i} * {j} = {i*j}")
    print()
````
