- Functions help in code reusability and better organization

````python
def isPrime(num):
    if num < 0:
        print("please provide positive number")
        return None
              
    if num == 1 or num == 0:
        return False
    for i in range(2,num):
        if num%i == 0:
            return False
    return True

if isPrime(0):
    print("Is prime")
else:
    print("Not a Prime")

# small bug try to find out
````
