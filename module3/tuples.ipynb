## Tuples
- a tuple is a built-in data type used to store an ordered, unchangeable (immutable), and indexed collection of items.

```python
# example 1
t1 = (10,20,30)
# or
t2 = 10,200,30


# what will be the answer
print(t1[0]) 
```

```python
# you can store any data types
# example

t = (10, 123.2, "CMRIT", [1,2,3], (6,7,8))
print(t)
```

```python
# Tuple Assignment
points = (1.2, 3.4, 6.5)
x, y, z = points
print("x value:",x)
print("y value:",y)
print("z value:",z)
```

```python
# create a funtion to find  min and max element present in the list
# And return the min and max value in tuple 
li = [-34,-35,-36, -34,-3, -2]

def min_max(li):
    
    mx = li[0]
    mi = li[0]
    for i in range(1,len(li)):
        if li[i] > mx:
            mx = li[i]
        if li[i] < mi:
            mi = li[i]
    return (mi, mx)
    
mi, mx = min_max(li)
print(mi)
print(mx)
```
