## Dictionaries
- A Python dictionary is a built-in data structure used to store data in key-value pairs
- The Key is what you use to find a specific piece of information.
- The Value is the actual data or information attached to that specific key.
- Dictionary are mutable 

```python
# To create a empty Dictionaries
vari = {}
```

### Dictionary operations
- deleting the data in  dictionary
- `del` keyword to delete by key
- len() length of the dictionary



```python
inventory = {
    "apples": 430, 
    "bananas": 312, 
    "oranges": 525, 
    "pears": 217
}
# deleting the pears from inventory 
del inventory["pears"]
print(inventory)

print(len(inventory))
```

### Dictionary methods

```python
inventory = {
    "apples": 430, 
    "bananas": 312, 
    "oranges": 525, 
    "pears": 217
}

# To access just all keys in form of list
print(list(inventory.keys()))

# accessing each keys from the dictionary though loops
for key in inventory:
    print(key)

# To access just all values in the dictionary
print(list(inventory.values()))

# To access both key and value in form of tuples
print(list(inventory.items()))
# To loop though key and value 
for key, val in inventory.items():
    print(key, val)


# To check whether key is present or not 
if "apples" in inventory:
    print("apple is present")
else:
    print("apple is not present")
```

### Aliasing and copying

```python
# Aliasing or reference
inventory = {
    "apples": 430, 
    "bananas": 312, 
    "oranges": 525, 
    "pears": 217
}
frts = inventory.copy()
fruits = inventory
fruits["apples"] = 12
print(inventory["apples"])


frts["apples"] = 234
print(inventory["apples"])
```

```python
# Counting letters , frequency of each letters
letter = "Mississippi"
letter_count = {}
for e in letter:
    if e in letter_count:
        letter_count[e]  = letter_count[e] + 1
    else:
        letter_count[e] = 1
        
# to display the result
for k , v in letter_count.items():
    print(f"{k} repeated {v} times")
    
    
```
