## 1. Explain the elif, or, and while statements in Python with a suitable example of each. [6 marks]



### `elif` Statement
The `elif` (else if) statement is used to check multiple conditions in sequence. It is used after an `if` and before an optional `else`.

**Example:**
```python

    print("Grade A")

    print("Grade B")
elif marks >= 60:
    print("Grade C")
else:
    print("Grade D")
```

### `or` Operator
The `or` operator is a logical operator that returns `True` if at least one of the conditions is `True`.

**Example:**
```python
age = 20
if age < 18 or age > 60:
    print("Special category")
else:
    print("Regular category")
```

### `while` Statement
The `while` statement is used for indefinite iteration. The loop continues as long as the condition remains `True`.

**Example:**
```python
count = 1
while count <= 5:
    print(count)
    count += 1
```

## 3. Define a user-defined function. How are parameters passed to functions? Explain with an example. [6 marks]



### User-defined Function
A user-defined function is a block of reusable code defined by the programmer using the `def` keyword. It performs a specific task and can be called multiple times.

### Parameter Passing
In Python, parameters are passed by **object reference** (also called pass-by-assignment). For immutable types (int, str, tuple), the function receives a copy of the reference. For mutable types (list, dict), modifications inside the function affect the original object.

**Example:**
```python
def greet(name, age):
    print(f"Hello {name}, you are {age} years old")

# Calling the function
greet("Alice", 25)

def modify_list(lst):
    lst.append(100)
    print("Inside function:", lst)

my_list = [1, 2, 3]
modify_list(my_list)
print("Outside function:", my_list)  # Original list is modified
```

## 8. Write a Python program to generate the Collatz 3n + 1 sequence until the value becomes 1. [8 marks]


````python
# Collatz 3n + 1 sequence
n = int(input("Enter a positive integer: "))
print("Collatz sequence:", end=" ")

while n != 1:
    print(n, end=" ")
    if n % 2 == 0:
        n = n // 2
    else:
        n = 3 * n + 1

print(1)  # Final value
````

## 20. Explain list aliasing and cloning with a suitable Python example. [8 marks]



### List Aliasing
When two variables refer to the **same list object** in memory. Changes made through one variable affect the other.

```python
list1 = [1, 2, 3]
list2 = list1          # Aliasing
list2.append(4)
print(list1)           # [1, 2, 3, 4]  (same object)
print(list1 is list2)  # True
```

### List Cloning
Creating a **new independent copy** of the list so that changes to one do not affect the other.

```python
list1 = [1, 2, 3]
list2 = list1[:]       # Cloning using slicing
# or list2 = list1.copy()
list2.append(4)
print(list1)           # [1, 2, 3]
print(list2)           # [1, 2, 3, 4]
print(list1 is list2)  # False
```

## 22. Develop a Python program to print the 10 most frequently appearing words in a text file. Use a dictionary to store distinct words and their frequencies, sort by frequency in descending order, and display the first 10 entries. [8 marks]


````python
# Note: Create a sample text file 'sample.txt' for testing
# Example content: This is a test. This is only a test.

from collections import Counter

def top_10_words(filename):
    with open(filename, 'r') as file:
        text = file.read().lower()
    
    # Remove punctuation and split into words
    words = ''.join(c if c.isalnum() or c.isspace() else ' ' for c in text).split()
    
    # Count frequencies
    word_count = Counter(words)
    
    # Get top 10
    top_10 = word_count.most_common(10)
    
    print("Top 10 most frequent words:")
    for word, count in top_10:
        print(f"{word}: {count}")

# Usage (uncomment and provide a file)
# top_10_words("sample.txt")
````

## 26. Write a Python program to read dictionary data and delete a specified key-value entry. [7 marks]


````python
# Program to read dictionary and delete a key
n = int(input("Enter number of entries: "))
d = {}

for i in range(n):
    key = input(f"Enter key {i+1}: ")
    value = input(f"Enter value {i+1}: ")
    d[key] = value

print("Original Dictionary:", d)

key_to_delete = input("Enter key to delete: ")
if key_to_delete in d:
    del d[key_to_delete]
    print("Updated Dictionary:", d)
else:
    print("Key not found!")
````

## 27. Explain how dictionaries are used as counters. Write a Python program to count the number of words in a text using a dictionary. [10 marks]



### Dictionaries as Counters
Dictionaries are commonly used as counters by storing elements as keys and their frequency as values. This is efficient for counting occurrences.

**Program:**


````python
text = input("Enter text: ")
words = text.split()

word_count = {}
for word in words:
    word = word.lower()
    word_count[word] = word_count.get(word, 0) + 1

print("Word frequencies:")
for word, count in word_count.items():
    print(f"{word}: {count}")
````

## 30. Write a Python program to merge two dictionaries. If a key occurs in both dictionaries, add their values. [8 marks]


````python
def merge_dicts(d1, d2):
    result = d1.copy()
    for key, value in d2.items():
        result[key] = result.get(key, 0) + value
    return result

# Example
dict1 = {"a": 10, "b": 20, "c": 30}
dict2 = {"b": 5, "c": 15, "d": 25}

merged = merge_dicts(dict1, dict2)
print("Merged Dictionary:", merged)
# Output: {'a': 10, 'b': 25, 'c': 45, 'd': 25}
````
